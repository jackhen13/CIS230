# a lambda function
**A Lambda function is a small anonymos function.*
**A Lambda function can take any number of arguments but can have one expression.

*Lambda Function Format: **Lambda arguments: expression

```python
def double_num(n):
    return n * 2

double_num(5)
```

# Higher-order functions, such as map(),filter(), and reduce()

**A function that operates with another function is known as a higher Order Functions.**
**It contains other functions as a parameter or returns a function as an output**

## A Map Function

**Python's map() is a built-in function that allows you to precess and transform all the items in an iterable without using explicit for loop**

**Map() is useful when you need to apply a transform function to each item in an iterable and transform them into a new iterable.**

**Map() function Format**

map(function, iterable)

```python
def square(number):
    return number ** 2

nums = list(range(1,16))

nums
```

```python
squares = [square(n) for n in nums]
squares
```

```python
#using Lambda
squares=list(map(square,nums))
squares
```

```python
import random

txns=random.choices(range(5,55), k=20)
print(txns)

TAX_RATE= .88

final_price = list(map(lambda n: n*(1+TAX_RATE), txns))
print('final_price:', final_price)
```
