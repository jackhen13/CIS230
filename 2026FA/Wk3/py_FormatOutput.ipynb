### Formatting Output with: %, .format(), and f-strings methods

```python
# 1 - Formatting output using the percent specifier sign %
print("The quick brown %s jumps over the lazy dog" % 'cat')
```

```python
print("The quick brown %s jump over the lazy %s" %('fox','dog'))
```

```python
#\t
print("The quick brown %s jumps over the lazy %s"%('\tfox','\tdog'))
```

```python
print("The quick brown %r jumps over the lazy%r"%('fox',"dog"))
```

```python
# Format Numbers
print("My score is %s in CIS310"%98.99)
```

```python
print("My score is %d is CIS310"%98.99)
```

```python
print("My score is %f is CIS310"%98.99)
```

```python
print("My score is %2.2f is CIS310"%98.99)
```

```python
print("My score is %2.1f is CIS310"%98.45)
```

```python
print("My score is %8.1f is CIS310"%98.45)
```

```python
# 2 - Formatted String Literals with the .format() Method
print("In CIS230 we use {} To preform data analysis" .format ('Python'))
```

```python
print("In CIS230 we use {!r} To preform data analysis" .format ('Python'))
```

```python
name='Python'
print("In CIS230 we use {} To preform data analysis" .format ('name'))
```

```python
print("In CIS230 we use {!r} To preform data analysis" .format ('name'))
```

```python
print("The {} {} jumps over the {} {}".format('brown','fox','lazy','dog'))
```

```python
print("The {2} {3} jumps over the {0} {1}".format('brown','fox','lazy','dog'))
```

```python
name1, name2 = 'fox', 'dog'
print("The {2} {3} jumps over the {0} {1}".format('brown',name1,'lazy',name2))
```

```python
# numbers
gpa = 3.67
print('my gpa is {0:6.1f}'.format(gpa))
```

```python
# 3 - Formatted String Literals f-Strings
name='Dr.Abdunabi'
term="Fall 2026"
course="CIS230"

print(f'{name} teaches {course} in {term}')
```
