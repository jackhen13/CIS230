### Python For Loops

**General for-loop format:**

```
for item in interchangable-object:
    statements to do work
```

````python
import random

sc= random.sample(range(50,101),k=15)
sc
````

````python
for n in sc:
    print(f'n={n**2}')
````

````python
for n in sc:
    if n % 2 ==0:
        print(f'n={n}')
````

````python
for n in sc:
    if n % 2 !=0:
        print(f'n= {n}')
````

````python
counter=0
for n in sc:
    if n % 2 ==0:
        print(f'n={n}')
        counter=counter+1
print(f'The total even numbers: {counter}')
````

````python
sc
````

````python
even=[]
odd=[]

for n in sc:
    if n % 2 ==0:
        even.append(n)
    else:
        odd.append(n)
````

````python
odd
````

````python
even
````

````python
s = 0

for n in sc:
    s+=n # s = s +n

print(f'The average of all scores is {s/15:2.2f}')
````

````python
for letter in "I am a student in CIS230, we are learning Python":
    if not letter.isspace():
        print(letter, end='')
    
````

````python
t=tuple(sc)
t
````

````python
colors = ['red','blue','green']
ratio = [200,199,89]

lst=list(zip(colors,ratio))
d = dict(lst)
d
````

````python
len(d)
````

````python
d.keys()
````

````python
d.values()
````

````python
d.items()
````

````python
for item in d:
    print(item)
````

````python
for key in d:
    print(d[key])
````

````python
for item in d.items():
    print(item[1])
````

````python
for k,v in d.items():
    print(k.v)
````

````python
#break and continue
fruits= ['Apple','Banana','Cherry','Orange']

for fruit in fruits:
    print(fruit)
    if fruit == 'Cherry':
        break
````

````python
for fruit in fruits:
    if fruit == 'Cherry':
        continue
    print(fruit)
````
