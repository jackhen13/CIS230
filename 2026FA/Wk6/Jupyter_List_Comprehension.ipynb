### Python List Comprhension 

**Python is famous for allowing you to write code that's elegant, easy to write, and almost as easy to read as plain english.**

**One of the languge's most distinctive features is the list comprehension, which you can use to create powerful functionality within a single line of code.**
## List Comprehension Format

1. new_list = [expression for member in iterable]
2. new_list = [expression for member in iterable **(if condishional**)
3. new_list = [expression **(if condishional**) for member in iterable]

```python
# create a list from a string

lst = []

for letter in 'hello':
    lst.append(letter)
lst
    
```

```python
# list comprehension
# new_list = [expression for member in iterable]

lst = [letter.upper() for letter in 'hello']
lst
```

```python
squares = []

for n in range(5,16):
    squares.append(n**2)
    
squares
```

```python
squares=[n**2 for n in range(5,16)]
squares
```

```python
# Convert Celsius to Farenheit
import random

c = random.sample(range(0,51), k=20)
c
```

```python
f=[((9/5)* t +32) for t in c]
f
```

```python
even = []

for t in c:
    if t % 2 ==0:
        even.append(t)
        
even
    
```

```python
# Filter the data, list comprehensions
# new_list = [expression for member in iterable (if condishional)

even=[t for t in c if t%2 == 0]
even
```

```python
txns = [1.09, 23.56, 57.84, 4.56,6.78]
TAX_RATE = .08

final_prices = [round(tx*(1+TAX_RATE),1) for tx in txns if tx > 10]
final_prices
```
