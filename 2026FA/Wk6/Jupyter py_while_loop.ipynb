### Python - While Loops
#### The general format of Python while loops

````
    while test:
    statements
````

`````python
num=5
while num <= 0:
    print(f"num={num}")
    num -= 1 # num - num - 1
`````

`````python
sum =0
n=10
i = 1

while i<=n:
    sum+=i
    i+=1
    
print("The sum of numbers from 1 to 10 is", sum)
`````

`````python
import random

lst = random.sample(range(1,101),k=10)
lst
`````

`````python
i = 0 

while i < len(lst):
    print(lst[i],'index:',i)
    i+=1
`````

`````python
for n in lst:
    print(n)
`````
