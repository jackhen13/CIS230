### Python Statements [if,else,elif]

**the format of the if-statement**

~~~
if condition1:
    perform action1
elif condition2:
    perform action2
else:
    perform action3
~~~

```python
value=int(input('Enter a number: '))

if value == 5:
    print('Hi Five!')
else:
    print("Lets try later!")
```

```python
#if-elif statement

score1 = float(input('Enter Test 1 Score: '))
score2 = float(input('Enter Test 2 Score: '))

if (score1 > score2):
    print(f"Test1:)'{score1}' is greater than Test2: '{score2}'")
    print("you mat think to redo Test2")
elif if (score1 > score2):
    print(f"Test1:)'{score2}' is greater than Test2: '{score1}'")
    print("you mat think to redo Test1")
else
    print(f"Test1: '{score1}'and Test2: '{score2}' scores are equal!")
    print('It is your choice to redo Test! or Test2')
```

```python
# One-Line if statements

d = {'name','Alice','age': 22, 'courses':['CIS230','CIS220','CIS350']}
if 'name'in d.keys(): print(d['name'])
```

```python
if 'Alice' in d.values(): print(d['name'])
```

```python
if 'CIS230' in d['courses']:indx=d['courses'].index('CIS230'); print(d['courses'][indx])
```

```python
# One-Line if statements inside on another function
raining=False
print("Lets head out for a",'run' if not raining else 'csu rec center')
```

```python
# An easy way to use built-in functions, instead of if-statement
names = {'Fred':'Hello Fred',
        'Joe':'Hello Joe',
         'Alice':'Hello Alice'
        }
names.keys()
```

```python
if 'Sam' in names.keys():
    print(names['Sam'])
else:
    print("I don't know who you are!")
```
