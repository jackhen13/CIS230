````python
# min_two_odds(3,5)--> 3
# min_two_odds(4,9)--> 9
def min_two_odds(n1,n2):
    if n1%2 !=0 and n2%2 !=0:
        return min(n1,n2)
    else:
        return max(n1,n2)
````

````python
min_two_odds(4,9)
````

````python
def is_prime(num):
    for n in range(2,num):
        if num % n == 0:
            return str(num)+ "is not prime."
    else:
        return str(num) + "is prime"
````

````python
is_prime(10)
````

````python
is_prime(11)
````

````python
def is_prime(num):
    for n in range(2,num):
        if num % n == 0:
            return False
    else:
        return True
````

````python
is_prime(10)
````

````python
is_prime(11)
````

````python
#check is a list has the number 5 next to the number 5
# somewhere in the list
#[3,4,5,5,10]->True
#[3,4,5,6,5,8]->False
def has_55(lst):
    for i in range(0,len(lst)-1):
        #if lst[i]== 5 and lst[i+1]==5:
        if lst[i:i+2]==[5,5]: 
            return True
        
    return False
````

````python
has_55([3,4,5,6,5,8])
````

````python
## Arbitray Keyword Arguments, **kwargs
# If you do not know how many keyword arguents that will be passed
# into your function, add two astrisk: # ** before the parameter name
# in the function defionition

def my_name(**names):
    print(names) # it is a dictionary
    if 'lname' in names and not 'fname' in names:
        print(f"Last name is {names['lname']}")
    elif  'fname' in names and 'lname' in names:
        print(f"first name is {names['fname']}")
    elif "fname" in names and "lname" in names:
        print(f"Full name is {names['lname']},{names['fname']}")
    else:
        print('No name is provided')
````

````python
my_name(fname="james",lname="Smith")# Jane Smith
````

````python
my_name()
````
