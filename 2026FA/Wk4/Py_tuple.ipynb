### Python Tuples (), tuple()

**Tuple**
- A tuple is a collection that is ordered and unchangable.
- Allows duplicate items with same values.
- Ordered - items have a defined order, index.
- Unchangable - Cannot change, add, ot remover items after the tuple is created.
