**Set**

- A set is a collection that is unordered (and unindexed)
- A set has no duplicate items.
- unordered - Items do not havea defined order, items cannot be referred to by index
- A set is unchangeable - Cannot change the items after the set is created, I can not update the item
- No Duplicates - No two items with the same value.
