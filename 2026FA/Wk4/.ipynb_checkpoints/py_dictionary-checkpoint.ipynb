### Python Dictionaries, {}, dict()

**Dictionary**

- A dictionary is a collection that is unordered and changeable. 
- Dictionaries store data values in *key: value* pairs.
- No Key Duplicates: Cannot have two items with the same key.
- Changeable - can change: add remove, update items after the dictionary is created.
- Item: key: value pairs

```python
d = {'key1':'value1','key2':'value2','key3':'value3','key4':'value1','key1':'value5'}
d
```
