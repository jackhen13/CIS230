# Data Types in NumPy

- i : integer
- b : boolean
- u : unsigned integer
- f : float
- c : complex float
- m : timedelta
- M : datetime
- O : object
- s : string
- u : unicode string
- v : fixed chunk of memory for other type (void)

# Checking the Data Type of an Array

- dtype : the returns the data type into the array
- astype : to cast an existing array data type to a specified data type(dtype)

```python
import numpy as np
import random
```
