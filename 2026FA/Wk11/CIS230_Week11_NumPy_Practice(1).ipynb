## Part 1: Building and Slicing Arrays (3 points)

A NumPy **array** holds a list of values that you can do math on all at once.

- `np.arange(n)` makes the whole numbers `0, 1, 2, ..., n-1`. It does the same job as `np.array(range(n))` from the lecture.
- `.reshape(rows, columns)` changes a flat array into a table. The total number of values must stay the same, so 10 values can become 2 rows of 5.
- The **ellipsis** `...` means "every value along the other dimensions." For a table, `table[..., 1]` means "column index 1 from every row." Remember that Python counts from 0, so column index 1 is the second column.

### Example (run this cell)

```python
# Example: run this cell and read the output. You do not need to change it.
example = np.arange(10)              # the numbers 0 to 9
print("example:", example)

example_2d = example.reshape(2, 5)   # 2 rows and 5 columns
print("example_2d:")
print(example_2d)
print("shape (rows, columns):", example_2d.shape)

print("example_2d[..., 1] gives column index 1 from every row:", example_2d[..., 1])
```

## Part 3: Best of Two Tests (3 points)

This part follows **Example 1.1: Computing the Higher Grade** from the lecture, which kept each student's better test score.

- `np.maximum(a, b)` compares two arrays position by position and keeps the **larger** value at each position.
- `np.minimum(a, b)` does the same but keeps the **smaller** value. It is handy for putting a ceiling on values, such as "no score above 100."
- `np.rint(a)` rounds each value to the nearest whole number, and `.astype(int)` turns the result into whole numbers (integers).
- Adding a number to an array, such as `a + 5`, adds it to **every** value.

### Example (run this cell)

```python
# Example: run this cell and read the output.
quiz_a = np.array([8.6, 6.2, 9.4])
quiz_b = np.array([7.9, 7.4, 9.1])

better = np.maximum(quiz_a, quiz_b)
print("Better quiz score:", better)                    # [8.6 7.4 9.4]

print("Rounded:", np.rint(better).astype(int))         # [9 7 9]

bonus = np.array([95, 102, 88])
print("Capped at 100:", np.minimum(bonus, 100))        # [ 95 100  88]
```

### Data for Part 3 (run this cell, do not change it)

```python
# Test 1 and Test 2 percentages for 5 students
test1 = np.array([67.4, 88.2, 45.9, 96.6, 73.1])
test2 = np.array([71.8, 84.5, 62.3, 90.4, 79.7])
print("test1:", test1)
print("test2:", test2)
```

### Your turn

```python
# Task 3a (1 point)
# For each student, keep the better of the two test scores.

best = np.maximum(test1, test2)

print("Best:", best)
```

```python
# Task 3b (1 point)
# Round best to the nearest whole number and convert it to integers.
# Hint: np.rint(...) first, then .astype(int) at the end.

best_rounded = np.rint(best).astype(int)

print("Best rounded:", best_rounded)
```

```python
# Task 3c (1 point)
# Give every student a 5 point curve, but no score may go above 100.
# Hint: add 5 to best_rounded, then put that inside np.minimum( , 100).

curved = np.minimum(best_rounded + 5, 100)

print("Curved:", curved)
```

## Part 4: Trend Line with `np.polyfit` (3 points)

This part follows **Example 2: Linear Regression** from the lecture. A straight trend line has the form

**y = slope × x + intercept**

- The **slope** is how much y changes when x goes up by 1.
- The **intercept** is the value of y when x is 0.
- `np.polyfit(x, y, 1)` finds the best straight line through the data. The `1` means a straight line. It returns the slope first and the intercept second, so you can write `slope, intercept = np.polyfit(x, y, 1)`.

You may see a value such as `9.999999999999996`. That is normal computer rounding and means 10.

### Example (run this cell)

```python
# Example from the lecture: study hours (X) and test scores (Y)
X = np.array([1, 2, 3, 4, 5, 6])
Y = np.array([50, 60, 70, 80, 90, 100])

ex_slope, ex_intercept = np.polyfit(X, Y, 1)
print("Slope:", round(ex_slope, 2))           # each extra hour adds about 10 points
print("Intercept:", round(ex_intercept, 2))   # predicted score with 0 hours of study

# Use the line to predict the score for 7 hours of study
print("Predicted score for 7 hours:", round(ex_slope * 7 + ex_intercept, 2))
```

## Part 5: Growth of an Investment (3 points)

This part follows **Example 3: Financial Calculations** from the lecture. With compound interest, the value after each year is

**future value = initial investment × (1 + rate) ^ years**

In Python, `^` is written as `**`. When `years` is an array, NumPy computes the future value for every year in one line.

`np.arange(start, stop)` counts from `start` and **stops before** `stop`. So `np.arange(1, 4)` gives `[1 2 3]`. The last item of an array is `a[-1]`.

### Example (run this cell)

```python
# Example from the lecture, shortened to 3 years
example_investment = 10000
example_rate = 0.05
example_years = np.arange(1, 4)    # [1 2 3]

example_value = example_investment * (1 + example_rate) ** example_years
for year, value in zip(example_years, example_value):
    print(f"Year {year}: ${value:,.2f}")

print("Value in the last year:", round(example_value[-1], 2))
```

### Data for Part 5 (run this cell, do not change it)

```python
initial_investment = 5000   # dollars
annual_rate = 0.04          # 4 percent per year
print("Investment:", initial_investment, " Rate:", annual_rate)
```

```python
# Task 5b (1 point)
# Compute the future value for every year in years.
# Hint: follow the formula above, using initial_investment, annual_rate, and years.

future_value = initial_investment * (1 + annual_rate) ** years

print("Future value:", future_value)
```

## Part 6: Portfolio, Success Rates, and Quality Control (4 points)

This part follows the finance, healthcare, and science examples at the end of the lecture notebook.

- **Portfolio return.** `np.dot(returns, weights)` multiplies each return by its weight and adds the results. For two assets, that is `returns[0] * weights[0] + returns[1] * weights[1]`.
- **Success rate.** When results are stored as 1 (success) and 0 (failure), `np.mean()` gives the share of successes. A mean of 0.6 means 60 percent succeeded.
- **Spread.** `np.std()` measures how far values usually fall from the mean. A small standard deviation means the values are consistent.

### Example (run this cell)

```python
# Examples from the lecture: run this cell and read the output.
ex_returns = np.array([0.07, 0.05])   # expected returns for Asset A and Asset B
ex_weights = np.array([0.6, 0.4])     # share of the money in each asset
print("Expected portfolio return:", round(np.dot(ex_returns, ex_weights), 4))

ex_group = np.array([1, 1, 0, 1, 0])  # 1 = success, 0 = failure
print("Success rate:", np.mean(ex_group))

ex_measurements = np.array([10.1, 10.5, 9.9, 10.3, 10.0])
print("Mean:", round(np.mean(ex_measurements), 4))
print("Standard deviation:", round(np.std(ex_measurements), 4))
```
